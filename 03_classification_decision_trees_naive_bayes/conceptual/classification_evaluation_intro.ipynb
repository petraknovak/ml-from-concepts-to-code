# Classification
In this notebook, we'll explore one of the most fundamental tasks in machine learning: classification. We'll focus on Decision Trees, a simple yet powerful model that can handle both numerical and categorical data. Our workflow will involve splitting the dataset into training and testing sets, training the model, making predictions, and evaluating its performance using common metrics. This hands-on example will give you a clear understanding of how machine learning models are trained, tested, and validated in a real-world scenario.


The first cell contains code that replicates what we did in Orange last week: it fits a Decision Tree model to the "zoo" dataset to predict the type of animal based on its features.

_Note: The code does not output anything._

```python
from pathlib import Path
import pandas as pd                                      # for data manipulation

from sklearn.model_selection import train_test_split     # for splitting the data
from sklearn.tree import DecisionTreeClassifier          # for the model

data_dir = Path("../../data")
data = pd.read_csv(data_dir / "zoo.csv")

X = data.drop(['type', 'name'], axis=1)                  # drop the target and the name from the features
y = data['type']                                         # set the target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)   # split the data into training and testing sets

model = DecisionTreeClassifier()              # initialize the model
model.fit(X_train, y_train)                   # fit the model  <--- this is where the "learning" of machine learning happens

y_hat = model.predict(X_test)                 # use the fitted (trained) model to make predictions on the test set
```

## Notation
In machine learning, we often use the following notation to represent various components of the problem:

__X__:
This represents the input data or features. It is typically a _matrix_ where each row corresponds to an example (instance) and each column corresponds to a feature (attribute) of the example. We use uppercase X because it is a two-dimensional array or matrix, following the convention in linear algebra where matrices are usually denoted by uppercase letters.

__y__:
This represents the target variable, which is the value we are trying to predict (the class in classification tasks). It is usually a _one-dimensional array or vector_ where each element corresponds to the target value for an example in X. Since y is a vector (one-dimensional), we use a lowercase letter.

__y_hat (ŷ)__:
This denotes the predicted target values. The "hat" symbol (ˆ) is commonly used in statistics and machine learning to represent an estimate or prediction. So y_hat is the model's predicted values for the corresponding features in X_test.

__train and test__:
train refers to the data used to train or fit the model. This includes X_train (features for training) and y_train (true target values for training).
test refers to the data used to evaluate the model's performance. It consists of X_test (features for testing) and y_test (true target values for testing), which the model has never seen during training. After training, we use the model to predict target values (y_hat) on the X_test set and compare them to y_test to evaluate model performance.

```python
# solution
def classification_accuracy(y, y_hat):
    """
    Computes the classification accuracy.
    
    Parameters:
        y (list or array): Actual class labels.
        y_hat (list or array): Predicted class labels.
        
    Returns:
        float: The classification accuracy as a percentage.
    """
    num_correct_predictions = sum(y == y_hat)
    num_predictions = len(y)
    return num_correct_predictions / num_predictions * 100
```

Let's compute the classification accuracy of our classifier.

```python
classification_accuracy(y_test, y_hat)   # In what percentage of cases do y_test and y_hat match?
```

## Example: Iris

The Iris dataset is a classic, beginner-friendly dataset often referred to as the "Hello World" of machine learning. 

The Iris dataset contains 150 samples of iris flowers, with each sample characterized by four features. These features describe the physical dimensions of the flowers, which belong to one of three species of iris:
Setosa, Versicolor, Virginica

Features (X):
- Sepal Length (cm) – The length of the outer part of the flower.
- Sepal Width (cm) – The width of the outer part of the flower.
- Petal Length (cm) – The length of the inner part of the flower.
- Petal Width (cm) – The width of the inner part of the flower.

Target (y):
- Species: 
Structure: one of Setosa, Versicolor, Virginica

```python
# import pandas as pd                                               # imported already above
# from sklearn.model_selection import train_test_split

# Load the Iris dataset (iris.csv) from a URL
url = "https://ifcs.boku.ac.at/repository/data/iris/iris.csv"
df = pd.read_csv(url)

df.head()
```

In scikit-learn, all data must be numerical for model training and evaluation. Therefore, we need to convert the categorical "Species" column into numerical format.

```python
# data preprocessing

species_map = {'setosa': 0, 'versicolor': 1, 'virginica': 2}

# transform the species column to integers  
df['Species'] = df['Species'].map(species_map)
```

```python
df
```

Fit and evaluate a decition tree model on the iris dataset.

```python
# Your code goes here

# Hints:
# 1. Drop the 'Species' column from the features  --> X
# 2. Set the target variable to the 'Species' column --> y
# 3. Split the data into training and testing sets  --> X_train, X_test, y_train, y_test
# 4. Initialize the model  (DecisionTreeClassifier)
# 5. Fit the model
# 6. Compute the classification accuracy by using model.score



```

```python
# Solution
X = df.drop('Species', axis=1)                  # drop the target from the features
y = df['Species']                               # set the target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)   # split the data into training and testing sets


model = DecisionTreeClassifier()                # initialize the model          
model.fit(X_train, y_train)                     # fit the model  

model.score(X_test, y_test)                     # scikit-learn has a built-in method for computing the classification accuracy
```

## Confusion matrix

```python
# What kind of errors does the model make?
# We need a confusion matrix to answer this question
from sklearn.metrics import confusion_matrix

y_hat = model.predict(X_test)                   # make predictions on the test data

# Compute confusion matrix
cm = confusion_matrix(y_test, y_hat)
print(cm)
```
