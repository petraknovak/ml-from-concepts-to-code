## Data leakage rule

**Major data leakage caps the total score at 40/100.**

Examples of major leakage:
- fitting an imputer, scaler, or encoder before the train/test split,
- calling `fit()` or `fit_transform()` on `X_test`,
- using `hotels_application.csv` to fit any preprocessing step,
- computing statistics from test/application data and using them in preprocessing.

Checking column names, shapes, or whether `transform()` succeeds is not leakage.
