```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

## The sigmoid function

```python
def sigma(z):
    """ Sigmoid function """
    return 1 / (1 + math.e**(-1*z))
```

```python
# Plot the sigmoid function
x = [i for i in np.linspace(-10, 10, 100)]
y = [sigma(i) for i in x]
plt.plot(x, y, lw=7)

plt.title('Sigmoid Function')
plt.xlabel('z')
plt.ylabel('sigma(z)')

# Draw Vertical asymptotes (z → -∞ and z → ∞)
plt.axvline(x=0, color='red', linestyle='--')

# Draw Horizontal asymptotes at y = 0 and y = 1
plt.axhline(y=0, color='blue', linestyle='--')
plt.axhline(y=1, color='blue', linestyle='--')

# Show plot
plt.show()
```

## Gradient of the Sigmoid Function

The sigmoid function, often used in binary classification tasks, is defined as:

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

### Properties of the Sigmoid Function

- The output of the sigmoid function ranges between 0 and 1.
- It has an S-shaped curve (sigmoidal) and is differentiable.

### Derivative of the Sigmoid Function

The derivative (or gradient) of the sigmoid function can be expressed as:

$$
\sigma'(x) = \sigma(x) \cdot (1 - \sigma(x))
$$

### Explanation of the Gradient

- This formula shows that the gradient depends on the output of the sigmoid function itself, $\sigma(x)$.
- When $\sigma(x)$ is close to 0 or 1, the gradient is small, indicating that the function is relatively flat in these regions. 
- When $\sigma(x)$ is around 0.5, the gradient is at its maximum, which means the function is steep and changes rapidly with respect to \(x\).

### Applications

The gradient of the sigmoid function is crucial for optimization algorithms like gradient descent, particularly in neural networks where it is used during the backpropagation step to update weights.



```python
def sigma_derivative(z):
    """ Derivative of the sigmoid function """
    sig = sigma(z)
    return sig * (1 - sig)

# Plot the derivative of the sigmoid function
x = [i for i in np.linspace(-10, 10, 100)]
y_derivative = [sigma_derivative(i) for i in x]
plt.plot(x, y_derivative, lw=7, color='orange')

plt.title('Derivative of Sigmoid Function')
plt.xlabel('z')
plt.ylabel('sigma\'(z)')

# Show plot
plt.show()
```
