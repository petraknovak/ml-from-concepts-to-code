## Example: Majority-class classifier


This classifier always predicts the most frequent class.

During `fit`, it stores the class labels and their frequencies. `predict` returns the majority class for every example, while `predict_proba` returns the learned class-frequency distribution for every example.

Notice how `fit` stores learned values on `self`, so that `predict` and `predict_proba` can use them later.

```python
import numpy as np


class Majority:
    def __init__(self):
        pass

    def fit(self, X, y):
        """Learn the classes and their distribution from the training data."""
        self.classes_, counts = np.unique(y, return_counts=True)
        self.class_probabilities_ = counts / counts.sum()
        self.majority_class_ = self.classes_[np.argmax(counts)]
        return self

    def predict(self, X):
        """Return the majority class for each example in X."""
        return np.full(len(X), self.majority_class_)

    def predict_proba(self, X):
        """Return the class distribution for each example in X."""
        return np.tile(self.class_probabilities_, (len(X), 1))
```

## Your tools
### `max_features`

In scikit-learn's `DecisionTreeClassifier`, the `max_features` parameter determines the maximum number of features considered when splitting a node. If it is an integer, that exact number of features is considered. If it is a float between 0 and 1, it specifies a fraction of the total number of features. For example, `max_features=0.5` means that half of the features are considered.

More details: https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html

### Random choices with `np.random`

NumPy's `np.random` module can make random choices. For example, `np.random.choice(n_samples, size=sample_size, replace=True)` selects row indices for a bootstrap sample. 
To make the results reproducible, use a seeded generator such as `rng = np.random.default_rng(random_seed)` and then call `rng.choice(...)`.
